```python
%sql
CREATE SCHEMA IF NOT EXISTS bankof420.bank_gold;

-- creating schema


```

### Customer Spend Analytics

```python
gold_customer_spend = (
    spark.table("bankof420.bank_cards_silver.card_transactions").alias("t")
    .join(
        spark.table("bankof420.bank_cards_silver.cards").alias("c"),
        "card_id"
    )
    .groupBy("customer_id")
    .sum("amount")
    .withColumnRenamed("sum(amount)", "total_spent")
)

gold_customer_spend.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("bankof420.bank_gold.customer_spend")
```

### Branch-wise Loan Exposure

```python
gold_branch_loans = (
    spark.table("bankof420.bank_loans_silver.loans")
    .groupBy("branch_id")
    .sum("loan_amount")
    .withColumnRenamed("sum(loan_amount)", "total_loan_amount")
)

gold_branch_loans.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("bankof420.bank_gold.branch_loan_summary")
```
